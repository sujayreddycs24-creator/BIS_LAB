```python

import random

# Parameters
population_size = 30
generations = 50
head_length = 3

functions = ['+', '-', '*']
terminals = ['x', '1', '2']

# Bigger input range
inputs = range(-10, 11)

# Tail length formula
tail_length = head_length + 1


# Generate random chromosome
def create_chromosome():
    head = [random.choice(functions + terminals)
            for i in range(head_length)]

    tail = [random.choice(terminals)
            for i in range(tail_length)]

    return head + tail


# Convert chromosome into expression
def express(chromosome):
    index = [0]

    def build():
        symbol = chromosome[index[0]]
        index[0] += 1

        if symbol in functions:
            left = build()
            right = build()
            return "(" + left + symbol + right + ")"

        return symbol

    return build()


# Calculate fitness
def fitness(chromosome):
    expression = express(chromosome)
    error = 0

    for x in inputs:
        try:
            predicted = eval(expression, {"x": x})
            actual = x * x + 1
            error += (actual - predicted) ** 2
        except:
            return 0

    return 1 / (1 + error)


# Mutation
def mutate(chromosome):
    child = chromosome.copy()

    for i in range(len(child)):
        if random.random() < 0.1:
            if i < head_length:
                child[i] = random.choice(functions + terminals)
            else:
                child[i] = random.choice(terminals)

    return child


# Initialize population
population = [create_chromosome()
              for i in range(population_size)]


# Main GEP algorithm
itr = 0
for generation in range(generations):
    itr += 1

    population.sort(key=fitness, reverse=True)

    best = population[0]

    # Print intermediate results
    if (itr % 10 == 0):
      print("\nGeneration:", generation + 1)
      print("Best Chromosome:", best)
      print("Best Expression:", express(best))
      print("Fitness:", fitness(best))

    # Keep the best chromosome
    new_population = [best]

    # Select and mutate chromosomes
    while len(new_population) < population_size:
        parent = random.choice(population[:10])
        child = mutate(parent)
        new_population.append(child)

    population = new_population


# Final result
best = max(population, key=fitness)

print("\n========== FINAL RESULT ==========")
print("Best Chromosome:", best)
print("Best Expression:", express(best))
print("Final Fitness:", fitness(best))

# Display predictions
print("\nInput | Predicted | Actual")

for x in inputs:
    predicted = eval(express(best), {"x": x})
    actual = x * x

    print(x, "   |", predicted, "      |", actual)
```
